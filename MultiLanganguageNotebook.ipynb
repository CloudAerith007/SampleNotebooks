```sos
from sklearn.datasets import load_iris
from sklearn import linear_model
import pandas as pd

iris = load_iris()
x = pd.DataFrame(iris.data, columns = ["SEPAL_LENGTH", "SEPAL_WIDTH", "PETAL_LENGTH", "PETAL_WIDTH"])
y = pd.DataFrame(list(map(lambda x: {0:'setosa', 1: 'versicolor', 2:'virginica'}[x], iris.target)), columns = ['SPECIES'])
iris_df = pd.concat([x,y], axis=1)
IRIS_TMP = oml.push(iris_df)
print("Shape:",IRIS_TMP.shape)
print("Columns:",IRIS_TMP.columns)
#z.show(IRIS_TMP.head(4))
display(IRIS_TMP.head(4))
```

```sos
library(ORE)

service_level <- tolower(trimws(Sys.getenv("SERVICE_LEVEL", unset = "")))
if (!nzchar(service_level)) {
  stop("Select High, Medium, or Low from the notebook toolbar, then rerun this cell.")
}

tns_path <- file.path(Sys.getenv("TNS_ADMIN", "/u01/wallet"), "tnsnames.ora")
lines <- readLines(tns_path, warn = FALSE)

aliases <- trimws(sub("=.*$", "", lines[grepl("=", lines, fixed = TRUE)]))
aliases <- aliases[!grepl("_drcp$", aliases, ignore.case = TRUE)]

service_alias <- aliases[
  tolower(sub("^.*_", "", aliases)) == service_level
]

if (length(service_alias) != 1L) {
  stop(sprintf(
    "Expected one alias for selected level '%s'; found: %s",
    service_level,
    if (length(service_alias)) paste(service_alias, collapse = ", ") else "<none>"
  ))
}

ore.connect(conn_string = service_alias[[1]])
ore.attach()
```
